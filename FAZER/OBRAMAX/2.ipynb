```python
import requests
from bs4 import BeautifulSoup

def coletar_links_obramax(url):
    headers = {
        "User-Agent": (
            "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
            "AppleWebKit/537.36 (KHTML, like Gecko) "
            "Chrome/115.0.0.0 Safari/537.36"
        )
    }

    response = requests.get(url, headers=headers)
    response.raise_for_status()

    soup = BeautifulSoup(response.text, "html.parser")

    links = []
    produtos = soup.select('a.vtex-product-summary-2-x-clearLink[href]')

    for produto in produtos:
        href = produto['href']
        if href.startswith("/"):
            links.append("https://www.obramax.com.br" + href)

    return links

# Exemplo de uso
if __name__ == "__main__":
    url = "https://www.obramax.com.br/pisos-e-revestimentos/pisos-ceramicos?page=1"
    links_produtos = coletar_links_obramax(url)

    print(f"🔗 Total de produtos encontrados: {len(links_produtos)}")
    for link in links_produtos:
        print(link)
```
