```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.common.exceptions import TimeoutException
import time

def iniciar_driver():
    options = Options()
    options.add_argument("--headless")  # Remova se quiser ver o navegador
    options.add_argument("--no-sandbox")
    options.add_argument("--disable-dev-shm-usage")
    return webdriver.Chrome(options=options)

def clicar_mostrar_mais(driver):
    try:
        botao = WebDriverWait(driver, 5).until(
            EC.element_to_be_clickable((By.XPATH, "//a[contains(text(), 'Mostrar mais')]"))
        )
        driver.execute_script("arguments[0].scrollIntoView(true);", botao)
        botao.click()
        return True
    except TimeoutException:
        return False

def esperar_carregamento(driver, quantidade_antes):
    try:
        WebDriverWait(driver, 10).until(
            lambda d: len(d.find_elements(By.CSS_SELECTOR, "a.vtex-product-summary-2-x-clearLink")) > quantidade_antes
        )
        return True
    except TimeoutException:
        return False

def coletar_todos_links(url):
    driver = iniciar_driver()
    driver.get(url)
    time.sleep(3)

    while True:
        produtos_atuais = driver.find_elements(By.CSS_SELECTOR, "a.vtex-product-summary-2-x-clearLink")
        quantidade_antes = len(produtos_atuais)

        if not clicar_mostrar_mais(driver):
            break

        # Scroll até o final para forçar renderização dos próximos produtos
        driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
        time.sleep(1.5)

        if not esperar_carregamento(driver, quantidade_antes):
            break

    print("✅ Todos os produtos carregados.")
    produtos_finais = driver.find_elements(By.CSS_SELECTOR, "a.vtex-product-summary-2-x-clearLink")
    links = list({p.get_attribute("href") for p in produtos_finais if p.get_attribute("href")})

    driver.quit()
    return links

# Execução
if __name__ == "__main__":
    url = "https://www.obramax.com.br/pisos-e-revestimentos/pisos-ceramicos"
    todos_links = coletar_todos_links(url)

    print(f"\n🔗 Total de links coletados: {len(todos_links)}")
    for link in todos_links:
        print(link)
```
