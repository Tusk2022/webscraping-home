```python
def get_product_data(url, headers, page_number):
    try:
        products_data = []
        params = {'page': page_number}
        response = requests.get(url, headers=headers, params=params)
        response.raise_for_status()  # Lança uma exceção se houver um erro HTTP
        soup = BeautifulSoup(response.text, "html.parser")
        products = soup.select('.category-products .product-layout')
        data_de_coleta = date.today()
        nome_empresa = "Fogões Shop"  # Substitua pelo nome correto da empresa

        for product in products:
            try:
                # Extrair informações do produto
                name = product.select_one('.item-title a').get_text(strip=True)
                price = product.select_one('.price').get_text(strip=True)
                image = product.select_one('.thumb img')['src']
                product_url = product.select_one('.item-title a')['href']

                product_data = {
                    "Nome empresa": nome_empresa,
                    "Data de coleta": data_de_coleta,
                    "Imagem": image,
                    "Nome": name,
                    "Preço": price,
                    "URL do produto": product_url
                }
                products_data.append(product_data)
            except Exception as e:
                print(f"Erro ao obter dados do produto: {e}")

        return products_data
        #ACUSADOR DE ERRO
    except Exception as e:
        print(f"\033[91mErro ao obter dados do produto: {e}\033[0m")
        return None
```
