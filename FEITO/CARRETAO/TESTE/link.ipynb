```python
import requests
from bs4 import BeautifulSoup
import time

# URL inicial da categoria
url = "https://carretaoconstrucao.com.br/pisos-e-revestimentos.html"

links_produtos = set()

while url:
    print(f"Coletando página: {url}")
    response = requests.get(url)
    response.raise_for_status()
    soup = BeautifulSoup(response.text, "html.parser")

    # Coletar links de produtos
    for a in soup.select("a.product-item-link"):
        href = a.get("href")
        if href and href.startswith("http"):
            links_produtos.add(href)

    # Procurar o link da próxima página
    next_link = soup.find("link", rel="next")
    if next_link:
        url = next_link.get("href")
        time.sleep(1)  # pausa para não sobrecarregar o servidor
    else:
        url = None

# Exibir resultados
print("\nLinks encontrados:")
for link in sorted(links_produtos):
    print(link)

print(f"\nTotal de produtos coletados: {len(links_produtos)}")
```
