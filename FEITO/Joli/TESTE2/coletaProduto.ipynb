```python
import requests
from bs4 import BeautifulSoup

# URL do produto
url = 'https://www.joli.com.br/revestimento-34x60-pei3-lc-avila-bege-hd-cx2-10m2-crie-um-novo-espaco-formigres/p'

# Cabeçalhos para simular um navegador
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/115.0.0.0 Safari/537.36'
}

# Requisição HTTP
response = requests.get(url, headers=headers)

# Verifica se a requisição foi bem-sucedida
if response.status_code == 200:
    soup = BeautifulSoup(response.text, 'html.parser')

    # Extração do nome do produto
    nome = soup.find('h1')
    nome_produto = nome.get_text(strip=True) if nome else 'Nome não encontrado'

    # Extração do preço
    preco = soup.find('span', class_='sales')
    preco_produto = preco.get_text(strip=True) if preco else 'Preço não encontrado'

    # Extração da referência
    referencia = soup.find('span', class_='skuReference')
    referencia_produto = referencia.get_text(strip=True) if referencia else 'Referência não encontrada'

    # Extração da descrição
    descricao = soup.find('div', class_='productDescription')
    descricao_produto = descricao.get_text(strip=True) if descricao else 'Descrição não encontrada'

    # Extração das especificações técnicas
    especificacoes = soup.find('div', class_='productSpecification')
    especificacoes_produto = especificacoes.get_text(strip=True) if especificacoes else 'Especificações não encontradas'

    # Exibição dos dados extraídos
    print(f'Nome do Produto: {nome_produto}')
    print(f'Preço: {preco_produto}')
    print(f'Referência: {referencia_produto}')
    print(f'Descrição: {descricao_produto}')
    print(f'Especificações Técnicas: {especificacoes_produto}')
else:
    print(f'Erro ao acessar a página: {response.status_code}')
```

```python
import requests
from bs4 import BeautifulSoup

def coletar_dados_produto(url):
    headers = {
        "User-Agent": "Mozilla/5.0"
    }
    resposta = requests.get(url, headers=headers)
    if resposta.status_code != 200:
        print(f"Erro ao acessar a página: {resposta.status_code}")
        return

    soup = BeautifulSoup(resposta.content, "html.parser")

    # Nome do produto
    nome_tag = soup.find("h1")
    nome = nome_tag.get_text(strip=True) if nome_tag else "Nome não encontrado"

    # Preço
    preco_tag = soup.find("div", class_="joli-joli-theme-1-x-price")
    preco = preco_tag.get_text(strip=True) if preco_tag else "Preço não encontrado"

    # Referência
    referencia_tag = soup.find("span", string=lambda text: text and "Referência:" in text)
    referencia = referencia_tag.get_text(strip=True).replace("Referência:", "").strip() if referencia_tag else "Referência não encontrada"

    # Descrição
    descricao_tag = soup.find("div", class_="joli-joli-theme-1-x-description")
    descricao = descricao_tag.get_text(strip=True) if descricao_tag else "Descrição não encontrada"

    # Especificações Técnicas
    especificacoes = {}
    ficha_tecnica_tag = soup.find("div", class_="joli-joli-theme-1-x-specifications")
    if ficha_tecnica_tag:
        linhas = ficha_tecnica_tag.find_all("div", class_="joli-joli-theme-1-x-specification")
        for linha in linhas:
            chave_tag = linha.find("span", class_="joli-joli-theme-1-x-specificationName")
            valor_tag = linha.find("span", class_="joli-joli-theme-1-x-specificationValue")
            if chave_tag and valor_tag:
                chave = chave_tag.get_text(strip=True)
                valor = valor_tag.get_text(strip=True)
                especificacoes[chave] = valor
    else:
        especificacoes = "Especificações não encontradas"

    # Exibir os dados coletados
    print(f"Nome do Produto: {nome}")
    print(f"Preço: {preco}")
    print(f"Referência: {referencia}")
    print(f"Descrição: {descricao}")
    print("Especificações Técnicas:")
    if isinstance(especificacoes, dict):
        for chave, valor in especificacoes.items():
            print(f"  {chave}: {valor}")
    else:
        print(especificacoes)

# URL do produto
url_produto = "https://www.joli.com.br/piso-61x61-ld-gales-hd-61-cx2-23m2--moderno-e-elegante-para-o-seu-ambiente-marcela/p"
coletar_dados_produto(url_produto)
```

```python
import requests
from bs4 import BeautifulSoup

url = "https://www.joli.com.br/revestimento-34x60-pei3-lc-avila-bege-hd-cx2-10m2-crie-um-novo-espaco-formigres/p"
headers = {"User-Agent": "Mozilla/5.0"}

response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text, "html.parser")

# Nome do produto - definir manualmente se não for encontrado automaticamente
nome_produto = "Revestimento 34x60 Pei3 Lc Avila Bege Hd Cx2.10m² - Formigres"

# Preços
preco_original = soup.select_one(".joli-joli-theme-1-x-list-price .joli-joli-theme-1-x-price")
preco_promocional = soup.select_one(".joli-joli-theme-1-x-selling-price .joli-joli-theme-1-x-price")

# Imagem
img_tag = soup.select_one("img.joli-store-components-0-x-productImageTag--main")
img_url = img_tag["src"] if img_tag else None

# Descrição
descricao_div = soup.select_one(".joli-joli-theme-1-x-product-description")
descricao_texto = descricao_div.get_text(separator="\n", strip=True) if descricao_div else None

# Exibindo os dados coletados
print(f"Produto: {nome_produto}")
print(f"Preço Original: {preco_original.text if preco_original else 'Não encontrado'}")
print(f"Preço Promocional: {preco_promocional.text if preco_promocional else 'Não encontrado'}")
print(f"Imagem: {img_url}")
print("\nDescrição:")
print(descricao_texto if descricao_texto else "Não encontrada")
```

```python
pip install selenium beautifulsoup4


from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.common.exceptions import NoSuchElementException
from bs4 import BeautifulSoup
import time

def coletar_dados_joli_selenium(url):
    # Configuração do navegador (modo headless opcional)
    options = Options()
    options.add_argument("--headless")
    options.add_argument("--no-sandbox")
    options.add_argument("--disable-dev-shm-usage")
    driver = webdriver.Chrome(options=options)

    try:
        driver.get(url)
        time.sleep(5)  # aguarda o carregamento da página com JavaScript

        soup = BeautifulSoup(driver.page_source, "html.parser")

        # Nome do Produto
        nome_tag = soup.find("h1")
        nome = nome_tag.get_text(strip=True) if nome_tag else "Nome não encontrado"

        # Preço Original
        preco_original_tag = soup.select_one(".joli-joli-theme-1-x-list-price .joli-joli-theme-1-x-price")
        preco_original = preco_original_tag.get_text(strip=True) if preco_original_tag else "Não encontrado"

        # Preço Promocional
        preco_promocional_tag = soup.select_one(".joli-joli-theme-1-x-selling-price .joli-joli-theme-1-x-price")
        preco_promocional = preco_promocional_tag.get_text(strip=True) if preco_promocional_tag else "Não encontrado"

        # Referência
        referencia_tag = soup.find("span", string=lambda text: text and "Referência:" in text)
        referencia = referencia_tag.get_text(strip=True).replace("Referência:", "").strip() if referencia_tag else "Referência não encontrada"

        # Descrição
        descricao_tag = soup.select_one(".joli-joli-theme-1-x-product-description")
        descricao = descricao_tag.get_text("\n", strip=True) if descricao_tag else "Descrição não encontrada"

        # Imagem
        imagem_tag = soup.select_one("img.joli-store-components-0-x-productImageTag--main")
        imagem_url = imagem_tag["src"] if imagem_tag else "Imagem não encontrada"

        # Especificações Técnicas
        especificacoes = {}
        ficha_tecnica_tag = soup.select_one(".joli-joli-theme-1-x-specifications")
        if ficha_tecnica_tag:
            especificacoes_tags = ficha_tecnica_tag.select(".joli-joli-theme-1-x-specification")
            for item in especificacoes_tags:
                chave = item.select_one(".joli-joli-theme-1-x-specificationName")
                valor = item.select_one(".joli-joli-theme-1-x-specificationValue")
                if chave and valor:
                    especificacoes[chave.text.strip()] = valor.text.strip()
        else:
            especificacoes = "Especificações não encontradas"

        # Exibir os dados
        print(f"\n🔹 Nome do Produto: {nome}")
        print(f"💰 Preço Original: {preco_original}")
        print(f"💸 Preço Promocional: {preco_promocional}")
        print(f"🆔 Referência: {referencia}")
        print(f"🖼️ Imagem: {imagem_url}")
        print("\n📄 Descrição:")
        print(descricao)
        print("\n📌 Especificações Técnicas:")
        if isinstance(especificacoes, dict):
            for chave, valor in especificacoes.items():
                print(f"  - {chave}: {valor}")
        else:
            print(especificacoes)

    finally:
        driver.quit()

# Exemplo de uso
url_produto = "https://www.joli.com.br/revestimento-34x60-pei3-lc-avila-bege-hd-cx2-10m2-crie-um-novo-espaco-formigres/p"
coletar_dados_joli_selenium(url_produto)
```
