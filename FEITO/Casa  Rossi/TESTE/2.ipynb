```python
# Instalações (Google Colab)
!apt-get update > /dev/null
!apt install chromium-chromedriver -y > /dev/null
!cp /usr/lib/chromium-browser/chromedriver /usr/bin
!pip install selenium pandas openpyxl > /dev/null


import requests
from bs4 import BeautifulSoup
from selenium import webdriver
from selenium.webdriver.common.by import By
import time


url_lista = 'https://www.casarossi.com.br/pisos-e-revestimentos/porcelanatos/acetinado'

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"
}
response = requests.get(url_lista, headers=headers)
soup = BeautifulSoup(response.text, 'html.parser')

links_produtos = []
produtos = soup.find_all('li', class_='item flex')

for produto in produtos:
    a_tag = produto.find('a', class_='space-image')
    if a_tag:
        link = a_tag['href']
        if not link.startswith("http"):
            link = "https://www.casarossi.com.br" + link
        links_produtos.append(link)

print(f"\nForam encontrados {len(links_produtos)} produtos.\n")


# Configuração do Selenium
options = webdriver.ChromeOptions()
options.add_argument('--headless')
options.add_argument('--no-sandbox')
options.add_argument('--disable-dev-shm-usage')
driver = webdriver.Chrome(options=options)

# Loop pelos links de produtos
for i, url in enumerate(links_produtos, start=1):
    print(f"\n== Produto {i} ==\nURL: {url}")
    try:
        driver.get(url)
        time.sleep(3)
        soup = BeautifulSoup(driver.page_source, 'html.parser')

        # Nome do produto
        try:
            nome = soup.find("h1", class_="fn").get_text(strip=True)
        except:
            nome = "N/A"

        # Preço por caixa
        try:
            preco_caixa = soup.find("span", class_="PrecoPrincipal").get_text(strip=True).replace("R$", "")
        except:
            preco_caixa = "N/A"

        # Preço por m²
        try:
            preco_m2 = soup.find("div", class_="preco-por-metro").get_text(strip=True)
        except:
            preco_m2 = "N/A"

        # m² por caixa
        try:
            m2_caixa = soup.find("div", class_="metro-na-caixa").get_text(strip=True)
        except:
            m2_caixa = "N/A"

        # Descrição
        try:
            descricao = ''
            for p in soup.find_all('p'):
                if 'Informações do Produto' in p.get_text():
                    descricao = p.get_text(separator='\n', strip=True)
                    break
        except:
            descricao = "N/A"

        # Imagem do produto
        try:
            img_tag = soup.find("div", class_="zoom").find("img")
            imagem_url = img_tag["src"]
        except:
            imagem_url = "N/A"

        # Exibir os dados
        print(f"Nome do Produto: {nome}")
        print(f"Preço por Caixa: R$ {preco_caixa}")
        print(f"Preço por m²: {preco_m2}")
        print(f"m² por Caixa: {m2_caixa}")
        print(f"Descrição:\n{descricao}")
        print(f"Imagem: {imagem_url}")

    except Exception as e:
        print(f"Erro ao processar o link: {e}")

# Encerra o driver
driver.quit()
```
