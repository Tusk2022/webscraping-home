```python
import requests
from bs4 import BeautifulSoup

# URL da página que você quer coletar os links
url = 'https://www.casarossi.com.br/pisos-e-revestimentos/porcelanatos/acetinado'

# Requisição HTTP
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"
}
response = requests.get(url, headers=headers)

# Parse do HTML
soup = BeautifulSoup(response.text, 'html.parser')

# Lista para armazenar os links
links_produtos = []

# Encontrar todos os produtos pelo <li class="item flex">
produtos = soup.find_all('li', class_='item flex')

for produto in produtos:
    a_tag = produto.find('a', class_='space-image')
    if a_tag:
        link = a_tag['href']
        links_produtos.append(link)

# Exibir os links coletados
for link in links_produtos:
    print(link)
```

```python
import requests
from bs4 import BeautifulSoup

def extrair_dados_produto(url):
    response = requests.get(url)
    response.raise_for_status()

    soup = BeautifulSoup(response.text, 'html.parser')

    # Nome do produto
    nome = soup.find('h1', class_='product-name')
    nome_produto = nome.get_text(strip=True) if nome else ''

    # Preço por metro
    preco_por_metro = soup.find('div', class_='preco-por-metro')
    preco_metro = preco_por_metro.get_text(strip=True) if preco_por_metro else ''

    # Preço por caixa
    preco_caixa_span = soup.find('span', {'id': 'variacaoPreco'})
    preco_caixa = preco_caixa_span.get_text(strip=True) if preco_caixa_span else ''

    # Quantos metros por caixa
    metro_na_caixa = soup.find('div', class_='metro-na-caixa')
    metros_caixa = metro_na_caixa.get_text(strip=True) if metro_na_caixa else ''

    # Código do produto (REF)
    codigo_ref_span = soup.find('span', id='product-reference')
    codigo_ref = codigo_ref_span.get_text(strip=True) if codigo_ref_span else ''

    # Descrição
    descricao = ''
    for p in soup.find_all('p'):
        if 'Informações do Produto' in p.get_text():
            descricao = p.get_text(separator='\n', strip=True)
            break

    # Imagem principal
    galeria = soup.find('div', class_='product-gallery')
    imagem_url = ''
    if galeria:
        img_tag = galeria.find('img')
        if img_tag and 'src' in img_tag.attrs:
            imagem_url = img_tag['src']

    # Montar dicionário com os dados
    dados = {
        'Nome do Produto': nome_produto,
        'Código do Produto': codigo_ref,
        'Preço por m²': preco_metro,
        'Preço por Caixa': preco_caixa,
        'm² por Caixa': metros_caixa,
        'Descrição': descricao,
        'Imagem': imagem_url,
        'URL': url
    }

    return dados

# Exemplo de uso
url_produto = 'https://www.casarossi.com.br/pisos-e-revestimentos/porcelanatos/acetinado/piso-formigres-66x66cm-acetinado-retificado-coimbra-2-18m'
dados = extrair_dados_produto(url_produto)

for k, v in dados.items():
    print(f"{k}: {v}")
```
