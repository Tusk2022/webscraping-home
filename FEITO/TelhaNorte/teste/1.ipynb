```python
# Instalação e configuração (caso ainda não tenha rodado no notebook)
!apt-get update > /dev/null
!apt install chromium-chromedriver -y > /dev/null
!cp /usr/lib/chromium-browser/chromedriver /usr/bin
!pip install selenium > /dev/null

# Imports
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
import time
import pandas as pd

# Configurações do navegador
options = Options()
options.binary_location = "/usr/bin/chromium-browser"
options.add_argument("--headless")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")

# Inicializa o navegador
driver = webdriver.Chrome(options=options)

# URL da categoria de porcelanatos
url = "https://www.telhanorte.com.br/pisos-e-revestimentos/porcelanatos"
driver.get(url)
time.sleep(5)  # aguarda carregar os produtos

# Rolar para carregar mais (pode repetir se quiser mais produtos)
driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
time.sleep(3)

# Seleciona os blocos de produto com base no HTML que você forneceu
produtos_elementos = driver.find_elements(By.CSS_SELECTOR, "div.vtex-search-result-3-x-galleryItem")

produtos = []

for produto in produtos_elementos:
    try:
        nome = produto.find_element(By.CSS_SELECTOR, "span.vtex-product-summary-2-x-productBrand").text.strip()
        preco = produto.find_element(By.CSS_SELECTOR, "span.telhanorte-telha-store-app-1-x-best-price-price-vitrini").text.strip()
        link = produto.find_element(By.CSS_SELECTOR, "a").get_attribute("href")
        produtos.append([nome, preco, link])
    except:
        continue

driver.quit()

# Salva em CSV
df = pd.DataFrame(produtos, columns=["Nome", "Preço", "URL"])
df.to_csv("telhanorte_produtos.csv", index=False)
df.head()
```
