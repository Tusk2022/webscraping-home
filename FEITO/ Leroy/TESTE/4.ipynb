```python
# Instalações (Colab)
!apt-get update > /dev/null
!apt install chromium-chromedriver -y > /dev/null
!cp /usr/lib/chromium-browser/chromedriver /usr/bin
!pip install selenium > /dev/null
!pip install pandas > /dev/null

# Imports
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.common.exceptions import NoSuchElementException, TimeoutException
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
import pandas as pd

# Setup Selenium
options = Options()
options.binary_location = "/usr/bin/chromium-browser"
options.add_argument("--headless")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")

driver = webdriver.Chrome(options=options)
wait = WebDriverWait(driver, 10)

# ========================
# PARTE 1: COLETAR LINKS (até 10)
# ========================

url = "https://www.leroymerlin.com.br/search?term=Revestimentos%20de%20Parede"
driver.get(url)

todos_links = set()
pagina = 1
LIMITE_PRODUTOS = 10

while len(todos_links) < LIMITE_PRODUTOS:
    print(f"🔄 Página {pagina}...")

    try:
        wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, "a[href*='/revestimento']")))
    except TimeoutException:
        print("⚠️ Timeout esperando os produtos.")
        break

    produtos = driver.find_elements(By.CSS_SELECTOR, "a[href*='/revestimento']")
    for p in produtos:
        link = p.get_attribute("href")
        if link:
            todos_links.add(link)
            if len(todos_links) >= LIMITE_PRODUTOS:
                break

    if len(todos_links) >= LIMITE_PRODUTOS:
        break

    try:
        botao_proximo = driver.find_element(By.CSS_SELECTOR, "button[rel='next']")
        if not botao_proximo.is_enabled():
            print("🚫 Botão 'Próxima' desabilitado. Fim da navegação.")
            break
        driver.execute_script("arguments[0].click();", botao_proximo)
        pagina += 1
    except NoSuchElementException:
        print("❌ Botão 'Próxima página' não encontrado.")
        break

print(f"\n✅ Total de links coletados: {len(todos_links)}\n")

# ========================
# PARTE 2: EXTRATOR DE DADOS DE CADA PRODUTO
# ========================

def extrair_dados_produto(url):
    driver.get(url)

    try:
        wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, "h1")))
    except TimeoutException:
        print(f"⚠️ Timeout esperando carregar: {url}")
        return {'URL': url, 'Erro': 'Timeout'}

    dados = {'URL': url}

    try:
        dados['Nome'] = driver.find_element(By.CSS_SELECTOR, "h1").text
    except:
        dados['Nome'] = None

    try:
        dados['Código'] = url.split("_")[-1]
    except:
        dados['Código'] = None

    # Preço Promocional: pega o valor + decimal
    try:
        preco_promo = driver.find_element(By.CSS_SELECTOR, '.css-dd8fqe-price-tag__container span[data-cy="product-price-integer"]').text
        preco_decimal = driver.find_element(By.CSS_SELECTOR, '.css-dd8fqe-price-tag__container span[data-cy="product-price-decimal"]').text
        dados['Preço Promocional'] = preco_promo + preco_decimal
    except:
        dados['Preço Promocional'] = None

    try:
        dados['Preço Original'] = driver.find_element(By.CSS_SELECTOR, 'p[data-cy="product-from-price"]').text
    except:
        dados['Preço Original'] = None

    try:
        embalagem = driver.find_element(By.CSS_SELECTOR, '.css-18uevif-price-tag__packaging-price strong').text
        area = driver.find_element(By.CSS_SELECTOR, '.css-18uevif-price-tag__packaging-price span:nth-of-type(2)').text
        dados['Preço por m²'] = f"{embalagem} | {area}"
    except:
        dados['Preço por m²'] = None

    try:
        dados['Descrição'] = driver.find_element(By.CSS_SELECTOR, 'meta[name="description"]').get_attribute("content")
    except:
        dados['Descrição'] = None

    try:
        dados['Imagem'] = driver.find_element(By.CSS_SELECTOR, 'meta[property="og:image"]').get_attribute("content")
    except:
        dados['Imagem'] = None

    try:
        especificacoes = {}
        linhas = driver.find_elements(By.CSS_SELECTOR, '.characteristics-table tr')
        for linha in linhas:
            try:
                chave = linha.find_element(By.CSS_SELECTOR, 'th').text.strip()
                valor = linha.find_element(By.CSS_SELECTOR, 'td').text.strip()
                especificacoes[chave] = valor
            except:
                continue
        dados['Especificações'] = especificacoes
    except:
        dados['Especificações'] = None

    dados['Loja'] = "Leroy Merlin"

    return dados

# ========================
# PARTE 3: LOOP SOBRE OS LINKS E COLETA DE DADOS
# ========================

todos_dados = []

for idx, link in enumerate(sorted(todos_links), 1):
    print(f"🔍 Coletando {idx}/{len(todos_links)}: {link}")
    try:
        dados = extrair_dados_produto(link)
        todos_dados.append(dados)
    except Exception as e:
        print(f"⚠️ Erro ao coletar {link}: {e}")

# ========================
# PARTE 4: SALVAR EM CSV
# ========================

df = pd.DataFrame(todos_dados)
df.to_csv("produtos_leroy.csv", index=False, encoding='utf-8-sig')

print("\n✅ Arquivo CSV gerado: produtos_leroy.csv")

driver.quit()
```
