```python
import requests
import pandas as pd

# Parâmetros da API do Algolia
ALGOLIA_APP_ID = "1CF3ZT43ZU"
ALGOLIA_API_KEY = "150c68d1c61fc1835826a57a203dab72"
ALGOLIA_INDEX = "production_products"

headers = {
    "Content-Type": "application/json",
    "X-Algolia-Application-Id": ALGOLIA_APP_ID,
    "X-Algolia-API-Key": ALGOLIA_API_KEY,
}

# Função para buscar por página
def buscar_porcelanatos(pagina=0):
    url = f"https://{ALGOLIA_APP_ID}-dsn.algolia.net/1/indexes/{ALGOLIA_INDEX}/query"
    payload = {
        "params": f"query=porcelanato&page={pagina}&hitsPerPage=100"
    }
    res = requests.post(url, headers=headers, json=payload)
    return res.json()

# Coleta paginada
produtos = []
pagina = 0

while True:
    dados = buscar_porcelanatos(pagina)
    hits = dados.get("hits", [])
    if not hits:
        break

    for item in hits:
        produtos.append({
            "Nome": item.get("name"),
            "Preço": item.get("price", {}).get("value"),
            "Preço Formatado": item.get("price", {}).get("formattedValue"),
            "Marca": item.get("brand"),
            "Link": f"https://www.leroymerlin.com.br{item.get('url')}",
            "Avaliação": item.get("rating", {}).get("average"),
            "Avaliações": item.get("rating", {}).get("count"),
            "Vendido por": item.get("seller", {}).get("name") or "Leroy Merlin"
        })

    print(f"Página {pagina + 1} coletada com {len(hits)} produtos")
    if pagina >= dados.get("nbPages", 0) - 1:
        break
    pagina += 1

# Salvar em CSV
df = pd.DataFrame(produtos)
df.to_csv("porcelanatos_leroy.csv", index=False, encoding="utf-8-sig")
print("Total coletado:", len(df))
df.head()
```
